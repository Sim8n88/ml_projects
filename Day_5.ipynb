```python
import pandas as pd
dane_sklepu = {
    'Produkt': ['Laptop', 'Telefon', 'Tablet', 'Monitor', 'Myszka', 'Klawiatura', 'Sluchawki'],
    'Cena': [2500, 1500, 1200, 800, 150, 300, 450],
    'Kategoria': ['IT', 'IT', 'IT', 'Biuro', 'Akcesoria', 'Akcesoria', 'Akcesoria'],
    'Magazyn': [5, 12, 0, 8, 30, 15, 0]
}
df = pd.DataFrame(dane_sklepu)
tanie_produkty=df['Cena']<1000
print(df[tanie_produkty])
Akcesoria=df['Kategoria']=='Akcesoria'
print(df[Akcesoria])
dwa_warunki=df[(df['Cena']>1000) & (df['Magazyn']>0)]
print(dwa_warunki)
zero_na_stanie=df['Magazyn']==0
print(df[zero_na_stanie])
print('#####')
drozsze_od_1000=df.loc[df['Cena']>1000,['Produkt']]
print(drozsze_od_1000)
Kategoria_IT=df.loc[df['Kategoria']=='IT',['Produkt','Magazyn']]
print(Kategoria_IT)
wynik=df.iloc[2:5,0:2]
print(wynik)
df.loc[df['Kategoria']=='Akcesoria',['Kategoria']]='Dodatki'
print(df)
```
