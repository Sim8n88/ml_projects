```python
import pandas as pd

sprzedaz = pd.DataFrame({
    'id_klienta': [1, 2, 1, 3, 2, 1],
    'kategoria': ['IT', 'AGD', 'IT', 'RTV', 'IT', 'AGD'],
    'kwota': [2500, 1200, 150, 3000, 800, 450]
})

klienci = pd.DataFrame({
    'id_klienta': [1, 2, 3, 4],
    'imie': ['Jan', 'Anna', 'Piotr', 'Maria']
})

suma_sprzedazy=sprzedaz.groupby('id_klienta')['kwota'].sum()
print(suma_sprzedazy)
srednia_sprzedaz=sprzedaz.groupby('id_klienta')['kwota'].mean()
print(srednia_sprzedaz)
laczna_ilosc_zamowien=sprzedaz.groupby('kategoria')['id_klienta'].count()
print(laczna_ilosc_zamowien)
pelna_tabela=pd.merge(sprzedaz,klienci,on='id_klienta')
print(pelna_tabela)
left_tabela=pd.merge(klienci,sprzedaz,on='id_klienta',how='left')
print(left_tabela)
dane_wynik=pelna_tabela.groupby('imie')['kwota'].sum()
print(dane_wynik)
malejaco=dane_wynik.sort_values(ascending=False)
print(malejaco)
jedna_linijka=pelna_tabela.groupby('imie')['kwota'].sum().sort_values(ascending=False)
print(jedna_linijka)
```
