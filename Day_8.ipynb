```python
import pandas as pd
import numpy as np

dane_zamowienia = {
    'id_klienta': [1, 2, 1, 3, 4, 1, 3, 2, np.nan, 4],
    'kategoria': ['AGD', 'IT', 'IT', 'RTV', 'IT', np.nan, 'RTV', 'AGD', 'IT', np.nan],
    'cena': [1500, 3000, 100, 2500, 4000, 200, 3000, np.nan, 800, 4500],
    'sztuki': [1, 2, 5, 1, 1, 3, np.nan, 1, 2, 1]
}
zamowienia = pd.DataFrame(dane_zamowienia)

dane_klienci = {
    'id_klienta': [1, 2, 3, 4, 5],
    'imie': ['Jan', 'Anna', 'Marek', 'Ewa', 'Tomasz'],
    'kraj': ['Polska', 'Polska', 'Niemcy', 'Polska', 'Niemcy']
}
klienci = pd.DataFrame(dane_klienci)
srednia_cena=zamowienia['cena'].mean()
zamowienia['cena']=zamowienia['cena'].fillna(srednia_cena)
zamowienia['sztuki']=zamowienia['sztuki'].fillna(1)
zamowienia_czyste=zamowienia.dropna().copy()
zamowienia_czyste['wartosc_zamowienia']=zamowienia['cena']*zamowienia_czyste['sztuki']
baza_glowna=pd.merge(zamowienia_czyste,klienci,on='id_klienta')
raport=baza_glowna.groupby('kraj')['wartosc_zamowienia'].sum()
print(raport)
raport_klientow=baza_glowna.groupby('imie')['wartosc_zamowienia'].sum().sort_values(ascending=False)
print(raport_klientow)
```
