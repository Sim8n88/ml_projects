```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
maile = pd.DataFrame({
    'Tresc': [
        'Wygrałeś milion dolarów, kliknij tutaj',
        'Spotkanie zespołu o 15:00 w sali',
        'Darmowy kredyt, super oferta, tylko dzisiaj',
        'Przesyłam fakturę za prąd i gaz',
        'Kliknij w link aby odebrać nagrodę',
        'Hej, idziemy dzisiaj na obiad?',
        'Pilne, twoje konto zablokowane, zaloguj się',
        'Raport finansowy jest gotowy do sprawdzenia'
    ],
    'Spam': [1, 0, 1, 0, 1, 0, 1, 0] # 1 = Spam, 0 = Zwykły mail (Cel)
})

Wektoryzator=CountVectorizer()
X=Wektoryzator.fit_transform(maile['Tresc'])
y=maile['Spam']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
model=RandomForestClassifier(random_state=42)
model.fit(X_train,y_train)
prognoza=model.predict(X_test)
wynik=accuracy_score(y_test,prognoza)
print(wynik)
nowa_wiadomosc = ["Hej, kliknij tutaj aby odebrać darmowy kredyt"]
nowa_wiadomosc__=Wektoryzator.transform(nowa_wiadomosc)
wynik2=model.predict(nowa_wiadomosc__)
print(wynik2)
```
