```python
import pandas as pd
import numpy as np

dane_pacjentow = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'wiek': [45, np.nan, 32, 55, np.nan],
    'cisnienie': [120, 130, np.nan, 140, 125],
    'wynik_testu': ['Pozytywny', 'Negatywny', 'Pozytywny', np.nan, 'Negatywny']
})
df = dane_pacjentow.copy() # Pracujemy na kopii, żeby łatwo było zacząć od nowa
print(df.isna().sum())
ostre_ciecie=df.dropna()
print(ostre_ciecie)
df['wynik_testu']=df['wynik_testu'].fillna('Negatywny')
print(df)
srednie_cisnienie=df['cisnienie'].mean()
print(srednie_cisnienie)
df['cisnienie']=df['cisnienie'].fillna(srednie_cisnienie)
print(df)
print(df['wynik_testu'].unique())
print(df['wynik_testu'].value_counts())
sredni_wiek=df['wiek'].mean()
print(sredni_wiek)
df['wiek']=df['wiek'].fillna(sredni_wiek)
print(df)
```
